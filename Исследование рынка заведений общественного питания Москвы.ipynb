# Документация

- Файл moscow_places.csv:
- name — название заведения;
- address — адрес заведения;
- category — категория заведения, например «кафе», «пиццерия» или «кофейня»;
- hours — информация о днях и часах работы;
- lat — широта географической точки, в которой находится заведение;
- lng — долгота географической точки, в которой находится заведение;
- rating — рейтинг заведения по оценкам пользователей в Яндекс Картах (высшая оценка — 5.0);
- price — категория цен в заведении, например «средние», «ниже среднего», «выше среднего» и так далее;
- avg_bill — строка, которая хранит среднюю стоимость заказа в виде диапазона, например:
- «Средний счёт: 1000–1500 ₽»;
- «Цена чашки капучино: 130–220 ₽»;
- «Цена бокала пива: 400–600 ₽».
- middle_avg_bill — число с оценкой среднего чека, которое указано только для значений из столбца avg_bill, начинающихся с подстроки «Средний счёт»:
- Если в строке указан ценовой диапазон из двух значений, в столбец войдёт медиана этих двух значений.
- Если в строке указано одно число — цена без диапазона, то в столбец войдёт это число.
- Если значения нет или оно не начинается с подстроки «Средний счёт», то в столбец ничего не войдёт.
- middle_coffee_cup — число с оценкой одной чашки капучино, которое указано только для значений из столбца avg_bill, начинающихся с подстроки «Цена одной чашки капучино»:
- Если в строке указан ценовой диапазон из двух значений, в столбец войдёт медиана этих двух значений.
- Если в строке указано одно число — цена без диапазона, то в столбец войдёт это число.
- Если значения нет или оно не начинается с подстроки «Цена одной чашки капучино», то в столбец ничего не войдёт.
- chain — число, выраженное 0 или 1, которое показывает, является ли заведение сетевым (для маленьких сетей могут встречаться ошибки):
- 0 — заведение не является сетевым
- 1 — заведение является сетевым
- district — административный район, в котором находится заведение, например Центральный административный округ;
- seats — количество посадочных мест.

# Создание столбца "is_24_7" 

Столбец будет включать обозначение ежедневного и круглосуточного режима работы заведения

В столбце необходимо отобразить, работает ли заведение круглосуточно и ежедневно или нет? Для реализации решения такой задачи подойдёт самописная функция

```python
def is_24_7(work): # создаём собственную функцию 
    if work == "ежедневно, круглосуточно":
        return True
    else:
        return False
```

# Исследование количества посадочных мест в заведениях по категориям

```python
seat_groupby = moscow_places.groupby(by = "category", as_index = False).agg(min = ("seats", "min"), max = ("seats", "max"), mean = ("seats", "mean"),median = ("seats", "median"))
# группируем заведения по посадочным местам
display(seat_groupby)
```

Можно разделилить заведения по категориям где разделителем будет количество посадочных мест. 
1. Заведение без посадочных мест (0) 
2. Заведение с малым количеством посадочных мест (0-30)
3. Заведение с средним количеством посадочных мест (31-80) 
4. Заведение с большим количеством посадочных мест (81-110) 
5. Заведение с огромным количеством посадочных мест (111+) 

```python
def type_of_seats(seats):
    try:
        if  seats == 0:
            return "0"
        elif 1 <= seats <= 30:
            return "1-30"
        elif 31 <= seats <= 80:
            return "31-80"
        elif 81 <= seats <= 110:
            return "81-110"
        elif seats >= 111:
            return "111+"
    except:
        pass
```

Создадим функцию принимающую на вход строку датасета с количеством посадочных мест

```python
moscow_places["seat_category"] = moscow_places["seats"].apply(type_of_seats) # применение функции для создания столбца
```

```python
display(moscow_places.head(20)) # проверим результат
```

В датасет добавлен столбец с указанием диапазона посадочных мест.

```python
data_for_groupby = moscow_places.groupby(["category", "seat_category"], as_index = False).agg(count = ("category", "count")).sort_values("count", ascending = False)
display(data_for_groupby) # подготовка таблицs для построения метрик
```

Сгруппируем данные по категориям и количеству посадочных мест.

```python
sns.catplot(x = "category", y = "count", hue = "seat_category", data = data_for_groupby, kind = "bar", height = 5, aspect = 2.5 )
plt.title("Категории посадочных мест в разновидностях заведений")
plt.xlabel("Категории")
plt.ylabel("Количество завдений")
plt.show()
```

Судя по графикам все посадочные категории встречаются во всех категориях заведениях. Очевидно что лидируют по количеству заведения с средним количеством посадочных мест, чуть менее часто встречаются заведения с огромным количеством посадочных мест. # Отображение всех заведений датасета на карте с помощью кластеров

```python
from folium import Map, Marker # импорт карты и маркера
from folium.plugins import MarkerCluster # импорт кластер

moscow_lat, moscow_lng = 55.751244, 37.618423 # задаём центр МСК

moscow_choro = Map(location = [moscow_lat, moscow_lng], zoom_start = 10, tiles = "Cartodb Positron")  # создаём пустой хороплет
create_cluster = MarkerCluster().add_to(moscow_choro)  # создаём пустой кластер

def born_cluster(row):  # для создания маркера в нужной точке воспользуемся самописной функцией 
    Marker(
        [row["lat"], row["lng"]],
        popup = f"{row['name']} {row['rating']}",
    ).add_to(create_cluster)  # добавляем в кластер

# Применяем функцию к каждому ряду 
moscow_places.apply(born_cluster, axis=1) 

moscow_choro  # вывод карты
```

Ранее построеные фоновые картограммы позволяют сформировать более точное понимание о распределении заведений по округам. На втором хороплете заведения объеденены в кластеры, что даёт возможность нам судить о плотности заведений в каждом конкрентном округе.
